## 3. Inspect the centrality results
These primary scores use the full 59-person graph. The separate main-component CSV recalculates them on 44 nodes.

```python
import csv
from html import escape
from IPython.display import HTML, Image, display

with open("results/centrality_all_nodes.csv") as stream:
    people = list(csv.DictReader(stream))
people.sort(key=lambda r: (-int(r["degree"]), int(r["node_id"])))
columns = ["node_id", "degree", "degree_centrality", "closeness", "betweenness", "eigenvector", "pagerank"]
head = "<tr>" + "".join(f"<th>{escape(c)}</th>" for c in columns) + "</tr>"
body = ""
for row in people[:10]:
    values = [row[c] if c in ["node_id", "degree"] else f"{float(row[c]):.4f}" for c in columns]
    body += "<tr>" + "".join(f"<td>{escape(v)}</td>" for v in values) + "</tr>"
display(HTML("<table>" + head + body + "</table>"))
```

## 4. View the figures
Component panels show the complete 59-person network. Community and clique figures focus on the main 44-person component. Layout distances are illustrative.

```python
for name in ["01_network_overview", "02_communities", "03_centrality_comparison", "04_maximum_clique", "05_distributions"]:
    display(Image(filename=f"figures/{name}.png", width=1000))
```

## 5. Inspect community sensitivity
Louvain seed 42 supplies the report figure. Seeds 0–29 test how the number of groups and modularity change. The ARI column measures chance-adjusted agreement with seed 42.

```python
with open("results/community_sensitivity.csv") as stream:
    runs = list(csv.DictReader(stream))
from collections import Counter
print("Community counts across seeds:", dict(Counter(int(r["communities"]) for r in runs)))
print("Modularity range:", min(float(r["modularity"]) for r in runs), max(float(r["modularity"]) for r in runs))
```
